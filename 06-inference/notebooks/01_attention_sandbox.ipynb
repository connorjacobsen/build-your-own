# Attention visibility sandbox

Optional companion to stages 02 and 04. This visualizes a small offset mask; it does not implement your attention function or grant stage credit. Change `cached` and `new_tokens` before predicting the visible cells.

```python
import numpy as np
import matplotlib.pyplot as plt
cached, new_tokens = 4, 3
queries = cached + np.arange(new_tokens)
keys = np.arange(cached + new_tokens)
visible = keys[None, :] <= queries[:, None]
plt.figure(figsize=(7, 3))
plt.imshow(visible, cmap="Blues", aspect="auto", vmin=0, vmax=1)
plt.yticks(range(new_tokens), queries)
plt.xticks(keys)
plt.xlabel("Absolute key position")
plt.ylabel("Absolute query position")
plt.title("Allowed attention positions")
plt.tight_layout()
plt.show()
```
