# Page-size and slack sandbox

Optional companion to stage 06. The chart models on-demand tail slack. Reservation-based admission adds another kind of slack, and resident pool bytes remain allocated even when blocks are free.

```python
import math
import matplotlib.pyplot as plt
lengths = [1, 7, 9, 17, 31]
block_sizes = [1, 2, 4, 8, 16]
slack = [sum(math.ceil(n/b)*b-n for n in lengths) for b in block_sizes]
plt.figure(figsize=(7, 3))
plt.bar([str(b) for b in block_sizes], slack, color="#2563eb")
plt.xlabel("Block size (tokens)")
plt.ylabel("Unused allocated tail slots")
plt.title("One source of cache fragmentation")
plt.tight_layout()
plt.show()
```
